# PIAACMC broadband — star + companion observation

Simulates a MagAO-X PIAACMC observation with two *incoherent* point sources:
the on-axis central star (suppressed by the coronagraph) and an off-axis
companion at a small angular separation, combined into one science image.

The simulation machinery is imported from `utils.piaacmc_bw` (the refactored
broadband code with the `cor_grid` fix).


````python
import numpy as np
import matplotlib.pyplot as plt

from utils.piaacmc_bw import observe_star_and_companion
````

## Observation parameters

- `SEPARATION` — companion separation in λ/D (sub-λ/D to a few λ/D is the PIAACMC sweet spot).
- `POSITION_ANGLE` — measured from +x. Keep `0` for the 1-D cut to pass through the companion.

The returned dict holds the broadband fields (normalised to the unocculted
on-axis star peak): `star` (suppressed residual), `companion` (off-axis PSF),
`science` (star + flux_ratio × companion) and `unocculted` (reference).

To reproduce the paper-3 binary detections, loop over separations/flux ratios:

